```python
import pandas as pd

df = pd.read_csv("data/AAPL.csv")
df.head()
```

```python
print("Dimensions :", df.shape)
print("\nTypes de colonnes :")
print(df.dtypes)
print("\nValeurs manquantes par colonne :")
print(df.isna().sum())
print("\nLignes en double :", df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])
print(df.dtypes)
```

```python
print("Trié par date croissante :", df["Date"].is_monotonic_increasing)
```

## Constat de l'inspection

- Aucune valeur manquante détectée sur les 7 colonnes
- Aucune ligne dupliquée
- Données déjà triées par date croissante
- Colonne `Date` convertie de texte vers un vrai type date (`datetime64`)

Aucune action corrective supplémentaire n'est nécessaire sur ces points.

```python
df["Variation"] = df["Close"].diff()
df["Variation_pct"] = df["Close"].pct_change() * 100
df[["Date", "Close", "Variation", "Variation_pct"]].head(10)
```
